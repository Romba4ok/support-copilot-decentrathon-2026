# Support Copilot: free GPU preflight
Only hardware/runtime metadata. No private data, package installation, model loading or training. Select a free standard GPU; do not buy compute or mount Drive.


```python
import sys, json, platform, pathlib, shutil, subprocess, importlib.metadata
from datetime import datetime, timezone

def version(name):
    try: return importlib.metadata.version(name)
    except importlib.metadata.PackageNotFoundError: return None
probe=subprocess.run(["nvidia-smi","--query-gpu=name,memory.total,memory.free,driver_version","--format=csv,noheader,nounits"],capture_output=True,text=True) if shutil.which("nvidia-smi") else None
disk=shutil.disk_usage("/content" if pathlib.Path("/content").exists() else ".")
report={"stage":"preflight_only","actual_training":False,"private_data_used":False,"python":sys.version.split()[0],"platform":platform.system(),"gpu_query_exit":probe.returncode if probe else None,"gpu":probe.stdout.strip() if probe else "unavailable","free_disk_gib":round(disk.free/1024**3,2),"preinstalled_torch":version("torch"),"preinstalled_transformers":version("transformers"),"timestamp_utc":datetime.now(timezone.utc).isoformat()}
print("SUPPORT_COPILOT_GPU_PREFLIGHT")
print(json.dumps(report,ensure_ascii=False,indent=2))
```
